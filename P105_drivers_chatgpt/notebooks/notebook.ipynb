# P105 - Conductores con ChatGPT

En esta actividad se repite el análisis de conductores de P103 (pandas) y P104 (SQLite), pero el código se le pidió a ChatGPT.

**Mensaje enviado a ChatGPT:**

> Tengo dos archivos CSV en la carpeta `../data/`:
> - `drivers.csv` con las columnas: driverId, name, ssn, location, certified, wage-plan
> - `timesheet.csv` con las columnas: driverId, week, hours-logged, miles-logged (un registro por conductor y por semana)
>
> Escribe código Python con pandas para un notebook de Jupyter que:
> 1. Cargue las dos tablas.
> 2. Calcule el total de horas (hours-logged) y de millas (miles-logged) de cada conductor.
> 3. Agregue el nombre (name) de cada conductor desde drivers.csv.
> 4. Guarde ese resumen en `../submission/summary.csv` sin el índice.
> 5. Haga un gráfico de barras horizontales con matplotlib de los 10 conductores con más millas y lo guarde en `../submission/top10_drivers.png`.
>
> Usa código sencillo, separa cada paso en una celda distinta y pon un comentario que explique qué hace cada una.

**Código que respondió ChatGPT:**

```python
# 1. Importar librerías y cargar los archivos CSV
import pandas as pd
import matplotlib.pyplot as plt
import os

drivers = pd.read_csv("../data/drivers.csv")
timesheet = pd.read_csv("../data/timesheet.csv")
```

```python
# 2. Calcular el total de horas y millas de cada conductor
resumen = timesheet.groupby("driverId", as_index=False).agg({
    "hours-logged": "sum",
    "miles-logged": "sum"
})
```

```python
# 3. Agregar el nombre de cada conductor
resumen = resumen.merge(
    drivers[["driverId", "name"]],
    on="driverId",
    how="left"
)

# Organizar las columnas del resumen
resumen = resumen[
    ["driverId", "name", "hours-logged", "miles-logged"]
]
```

```python
# 4. Crear la carpeta de salida y guardar el resumen sin índice
os.makedirs("../submission", exist_ok=True)
resumen.to_csv("../submission/summary.csv", index=False)
```

```python
# 5. Crear y guardar el gráfico de los 10 conductores con más millas
top10 = resumen.nlargest(10, "miles-logged").sort_values("miles-logged")

plt.figure(figsize=(10, 6))
plt.barh(top10["name"], top10["miles-logged"])

plt.xlabel("Total de millas")
plt.ylabel("Conductor")
plt.title("Top 10 conductores con más millas")
plt.tight_layout()

plt.savefig("../submission/top10_drivers.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Confirmar que los archivos fueron generados
print("Archivos guardados correctamente:")
print("../submission/summary.csv")
print("../submission/top10_drivers.png")
```
